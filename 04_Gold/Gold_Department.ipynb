```python
# ============================================================
# Read Silver Department Table
# ============================================================

silver_departments_df = read_delta(SILVER_DEPARTMENTS)

preview(
    silver_departments_df,
    "Silver Departments"
)
```

```python
# ============================================================
# Create Department Dimension
# ============================================================

gold_departments_df = (
    silver_departments_df
        .select(
            "department_id",
            "department_name",
            "division",
            "region",
            "cost_center",
            "annual_budget_usd"
        )
)
```

```python
# ============================================================
# Add Gold Metadata
# ============================================================

gold_departments_df = (
    gold_departments_df
        .withColumn(
            "gold_load_timestamp",
            current_timestamp()
        )
)
```

```python
# ============================================================
# Write Gold Delta Table
# ============================================================

write_delta(
    df=gold_departments_df,
    table_name=GOLD_DEPARTMENTS
)
```

```python
# ============================================================
# Validate Gold Delta Table
# ============================================================

validate_table(GOLD_DEPARTMENTS)
```
