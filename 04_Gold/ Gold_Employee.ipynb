```python
# ============================================================
# Read Silver Employee Table
# ============================================================

silver_employees_df = read_delta(SILVER_EMPLOYEES)

preview(silver_employees_df,"Silver Employees")
```

```python
# ============================================================
# Create Employee Dimension
# ============================================================

gold_employees_df = (
    silver_employees_df
        .select(
            "employee_id",
            "employee_name",
            "department_id",
            "job_title",
            "email",
            "hire_date",
            "is_active"
        )
)
```

```python
# ============================================================
# Add Gold Metadata
# ============================================================

gold_employees_df = (
    gold_employees_df
        .withColumn(
            "gold_load_timestamp",
            current_timestamp()
        )
)
```

```python
# ============================================================
# Write Gold Delta Table
# ============================================================

write_delta(
    df=gold_employees_df,
    table_name=GOLD_EMPLOYEES
)
```

```python
# ============================================================
# Validate Gold Delta Table
# ============================================================

validate_table(GOLD_EMPLOYEES)
```
